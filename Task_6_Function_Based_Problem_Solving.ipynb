### Problem 1: Smart Parking Lot Management System

```python
# Problem 1: Smart Parking Lot Management System

def smart_parking_system(capacity, vehicle_logs):
    current_count = 0
    peak_usage = 0
    
    for log in vehicle_logs:
        if log == "IN":
            current_count += 1
        elif log == "OUT" and current_count > 0:
            current_count -= 1
        
        # Track peak usage
        if current_count > peak_usage:
            peak_usage = current_count
    
    # Parking Status
    if current_count >= capacity:
        status = "Full"
    else:
        status = "Available"
    
    return current_count, peak_usage, status


capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

current, peak, parking_status = smart_parking_system(capacity, logs)

print("Currently Parked Vehicles:", current)
print("Peak Usage:", peak)
print("Parking Status:", parking_status)
```

### Problem 2: Online Food Delivery Time Estimator

```python
# Problem 2: Online Food Delivery Time Estimator

def estimate_delivery_time(distance, traffic, weather):
    base_time = distance * 5  # 5 minutes per km
    
    # Traffic adjustment
    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 8
    
    # Weather adjustment
    if weather == "Rainy":
        base_time += 10
    elif weather == "Stormy":
        base_time += 20
    
    return base_time


eta = estimate_delivery_time(8, "High", "Rainy")

print("Estimated Delivery Time:", eta, "minutes")
```

### Problem 3: Movie Theatre Seat Occupancy Analyzer

```python
# Problem 3: Movie Theatre Seat Occupancy Analyzer

def analyze_seat_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy_percentage = (booked_count / total_seats) * 100
    
    if occupancy_percentage == 100:
        status = "Housefull"
    elif occupancy_percentage >= 75:
        status = "Almost Full"
    else:
        status = "Seats Available"
    
    return occupancy_percentage, status


total_seats = 200
booked = [1] * 150  # 150 seats booked

occupancy, show_status = analyze_seat_occupancy(total_seats, booked)

print("Occupancy:", occupancy, "%")
print("Show Status:", show_status)
```

### Problem 4: Cloud Server Load Classification System

```python
# Problem 4: Cloud Server Load Classification System

def classify_server_load(cpu_readings):
    total = 0
    
    for reading in cpu_readings:
        total += reading
    
    average_cpu = total / len(cpu_readings)
    
    if average_cpu < 50:
        status = "Normal"
    elif 50 <= average_cpu <= 80:
        status = "Warning"
    else:
        status = "Critical"
    
    return average_cpu, status


cpu_data = [45, 60, 70, 85, 90]

avg_cpu, server_status = classify_server_load(cpu_data)

print("Average CPU Load:", avg_cpu, "%")
print("Server Status:", server_status)
```

### Problem 5: Smart Classroom Resource Usage Monitor

```python
# Problem 5: Smart Classroom Resource Usage Monitor

def monitor_resource_usage(resource_dict):
    overused = []
    
    for resource, hours in resource_dict.items():
        if hours > 8:  # Threshold for overuse
            overused.append(resource)
    
    energy_alert = "Yes" if overused else "No"
    
    return overused, energy_alert


resources = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

overused_resources, alert_status = monitor_resource_usage(resources)

print("Overused Resources:", overused_resources)
print("Energy Alert:", alert_status)
```

### Problem 6: Online Event Registration Capacity Controller

```python
# Problem 6: Online Event Registration Capacity Controller

def manage_event_registration(capacity, registrations):
    if registrations <= capacity:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    else:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    
    return confirmed, waitlisted, status


confirmed_users, waitlist_users, registration_status = manage_event_registration(100, 105)

print("Confirmed Registrations:", confirmed_users)
print("Waitlisted Users:", waitlist_users)
print("Registration Status:", registration_status)
```
