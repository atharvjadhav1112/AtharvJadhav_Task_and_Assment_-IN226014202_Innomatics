Problem 1: Unique Words in a Sentence

```python
# Problem 1: Unique Words in a Sentence

def unique_words(sentence):
    # Split sentence into words
    words = sentence.split()
    
    # Convert list to set to remove duplicates
    unique = set(words)
    
    # Display results
    print("Unique words count:", len(unique))
    print("Unique words:", unique)


# Example Execution
sentence = "python is easy and python is powerful"
unique_words(sentence)
```

# Problem 2: Highest Salary from Employee Data

```python
# Problem 2: Highest Salary from Employee Data

def highest_salary(employees):
    # Initialize max salary and employee
    max_employee = None
    max_salary = 0
    
    # Loop through dictionary
    for name, salary in employees.items():
        if salary > max_salary:
            max_salary = salary
            max_employee = name
    
    print(f"Highest Salary: {max_employee} - {max_salary}")


# Example Execution
employees = {
    "Ravi": 75000,
    "Anita": 68000,
    "Kiran": 72000
}

highest_salary(employees)
```

# Problem 3: Find Maximum and Minimum Values

```python
# Problem 3: Find Maximum and Minimum Values

def find_max_min(numbers):
    # Initialize max and min with first element
    maximum = numbers[0]
    minimum = numbers[0]
    
    # Compare each element
    for num in numbers:
        if num > maximum:
            maximum = num
        if num < minimum:
            minimum = num
    
    print("List:", numbers)
    print("Max:", maximum)
    print("Min:", minimum)


# Example Execution
numbers = [45, 22, 89, 10, 66]
find_max_min(numbers)
```

# Problem 4: Count Products Above 1000

```python
# Problem 4: Count Products Above 1000

def count_expensive_products(prices):
    count = 0
    
    for price in prices:
        if price > 1000:
            count += 1
    
    print("Products above 1000:", count)


# Example Execution
prices = [450, 1200, 899, 1500, 300]
count_expensive_products(prices)
```

# Problem 5: Calculate Attendance Percentage

```python
# Problem 5: Calculate Attendance Percentage

def attendance_percentage(attendance):
    total_days = len(attendance)
    present_days = attendance.count("P")
    
    percentage = (present_days / total_days) * 100
    
    print("Attendance Percentage:", percentage)


# Example Execution
attendance = ["P", "P", "A", "P", "P"]
attendance_percentage(attendance)
```

# Problem 7: Count Character Frequency

```python
# Problem 7: Count Character Frequency

def character_frequency(text):
    frequency = {}
    
    for char in text:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    
    print(frequency)


# Example Execution
text = "python"
character_frequency(text)
```

# Problem 9: Check if Key Exists in Dictionary

```python
# Problem 9: Check if Key Exists in Dictionary

def check_key(dictionary, key):
    if key in dictionary:
        print("Employee exists")
    else:
        print("Employee does not exist")


# Example Execution
employees = {"Ravi": 75000, "Anita": 68000}
check_key(employees, "Ravi")
```

# Problem 10: Calculate Average Marks

```python
# Problem 10: Calculate Average Marks

def average_marks(marks):
    total = sum(marks)
    average = total / len(marks)
    
    print("Average Marks:", average)


# Example Execution
marks = [70, 75, 80, 65]
average_marks(marks)
```
