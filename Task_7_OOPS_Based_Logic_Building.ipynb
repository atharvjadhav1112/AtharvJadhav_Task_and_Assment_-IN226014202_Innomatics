## Problem 1: Smart Light Controller

```python
# Problem 1: Smart Light Controller

class SmartLight:
    
    def __init__(self, light_name):
        self.light_name = light_name
        self.status = "OFF"   # Default state
    
    def turn_on(self):
        self.status = "ON"
    
    def turn_off(self):
        self.status = "OFF"
    
    def display_status(self):
        print(f"{self.light_name} is {self.status}")


# Sample Execution
light = SmartLight("Bedroom Light")
light.turn_on()
light.display_status()
```

## Problem 2: Employee ID Card System

```python
# Problem 2: Employee ID Card System

class EmployeeIDCard:
    
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    
    def display_id_card(self):
        print("Employee ID Card")
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)


# Sample Execution
employee = EmployeeIDCard("Rahul", "EMP102", "IT")
employee.display_id_card()
```

## Problem 3: Mobile Contact Record

```python
# Problem 3: Mobile Contact Record

class Contact:
    
    def __init__(self, contact_name, phone_number):
        self.contact_name = contact_name
        self.phone_number = phone_number
    
    def save_contact(self):
        print("Contact Saved")
        print("Name:", self.contact_name)
        print("Phone:", self.phone_number)


# Sample Execution
contact = Contact("Anita", "9876543210")
contact.save_contact()
```

##  Problem 4: Product Price Tag Generator

```python
# Problem 4: Product Price Tag Generator

class Product:
    
    def __init__(self, product_name, price):
        self.product_name = product_name
        self.price = price
    
    def generate_price_tag(self):
        print("Product:", self.product_name)
        print("Price: ₹" + str(self.price))


# Sample Execution
item = Product("Headphones", 2499)
item.generate_price_tag()
```

## Problem 5: Movie Rating Display System

```python
# Problem 5: Movie Rating Display System

class Movie:
    
    def __init__(self, movie_name, rating):
        self.movie_name = movie_name
        self.rating = rating
    
    def display_movie_info(self):
        print("Movie:", self.movie_name)
        print("Rating:", self.rating, "/ 5")


# Sample Execution
movie = Movie("Inception", 4.8)
movie.display_movie_info()
```

##  Problem 6: Delivery Address Manager

```python
# Problem 6: Delivery Address Manager

class DeliveryDetails:
    
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    
    def print_delivery_details(self):
        print("Delivery Details")
        print("Customer:", self.customer_name)
        print("Address:", self.address)


# Sample Execution
delivery = DeliveryDetails("Suman", "Hyderabad")
delivery.print_delivery_details()
```
